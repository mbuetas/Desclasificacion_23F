```python
import pandas as pd
import networkx as nx
import itertools

# 1. Cargar el dataset
df = pd.read_csv(r'C:\Users\migue\OneDrive\Escritorio\Machine-Learning_23F\dataset_23f.csv')

# 2. Limpieza de la columna 'actores'
# Quitamos los valores nulos (NaN)
actores_limpios = df['actores'].dropna()

# Convertimos cada texto (ej: "Tejero | Armada") en una lista de Python (["Tejero", "Armada"])
# Usamos strip() para quitar espacios en blanco sobrantes al principio o final de los nombres
listas_de_actores = actores_limpios.apply(lambda x: [actor.strip() for actor in x.split('|')])

# 3. Inicializar el Grafo
# Usamos un Grafo no dirigido (la relación va en ambos sentidos: si A habla con B, B habla con A)
G = nx.Graph()

# 4. Poblar el Grafo (Crear Nodos y Aristas)
for lista in listas_de_actores:
    # Si en el documento solo hay 1 actor, no hay conexión que dibujar
    if len(lista) > 1:
        # itertools.combinations crea todas las parejas posibles dentro de esa lista
        parejas = itertools.combinations(lista, 2)
        
        for origen, destino in parejas:
            # Si la conexión ya existe, le sumamos +1 al "peso" (aparecen juntos más de una vez)
            if G.has_edge(origen, destino):
                G[origen][destino]['weight'] += 1
            else:
                # Si es la primera vez que aparecen juntos, creamos la línea con peso 1
                G.add_edge(origen, destino, weight=1)

# 5. Comprobación inicial
print("--- RESUMEN DE LA RED DEL 23F ---")
print(f"Total de Actores distintos (Nodos): {G.number_of_nodes()}")
print(f"Total de Conexiones únicas (Aristas): {G.number_of_edges()}")
```
